```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Demo").getOrCreate()

spark
```

```python
jdbc_url = "jdbc:postgresql://postgres:5432/"
properties = {
    "user": "admin",
    "password": "admin",
    "driver": "org.postgresql.Driver"
}

# jdbc_url
```

```python
postgres_db = "spark_demo"
jdbc_url = jdbc_url + postgres_db
# jdbc_url
```

```python
from pyspark.sql import Row

# Sample DataFrame
data = [Row(name="Alice", email="alice@example.com", age=30),
        Row(name="Bob", email="bob@example.com", age=25)]
cust_df = spark.createDataFrame(data)
```

```python
cust_df.write.jdbc(
    url=jdbc_url,
    table="customers",
    mode="append",   # or "overwrite"
    properties=properties
)
```

```python
df_customers = spark.read.jdbc(
    url=jdbc_url,
    table="customers",
    properties=properties
)

df_customers.show()
```

```python
# Create DataFrame
data = [(1, "James", 30, "M"), (2, "Ann", 40, "F")]
columns = ["id", "name", "age", "gender"]
sampleDF = spark.sparkContext.parallelize(data).toDF(columns)
```

```python
sampleDF.write.jdbc(
    url=jdbc_url,
    table="public.person",
    mode="overwrite",
    properties=properties
)
```
